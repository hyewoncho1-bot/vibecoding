```
def calculate_stats(numbers):
    avg = sum(numbers) / len(numbers)
    maximum = max(numbers)
    minimum = min(numbers)

    print(f"--- 결과 ---")
    print(f"숫자 리스트: {numbers}")
    print(f"평균: {avg}")
    print(f"최댓값: {maximum}")
    print(f"최솟값: {minimum}")

# 실행 확인용 데이터
my_list = [10, 20, 30, 40, 50]
calculate_stats(my_list)
```

역할(Role): 너는 파이썬 데이터 분석 전문가야.

맥락(Context): 나는 매일 내가 쓴 돈을 기록하고 관리하고 싶어. 복잡한 앱보다는 내가 직접 실행해서 확인할 수 있는 간단한 프로그램을 만들려고 해.

목표(Goal): 이번 학기 끝날 때까지 '날짜, 항목, 금액'을 입력하면 파일로 저장하고, 한 달 총액을 보여주는 프로그램을 만들 거야. (MVP)

제약(Constraints): 파이썬(Python) 언어를 사용하고, 별도의 설치 없이 구글 코랩에서 바로 돌아가야 해. 결과는 표 형태로 깔끔하게 보여줘.

```
# 지출 기록을 저장할 리스트
expenses = []

def add_expense(date, item, amount):
    expenses.append({"날짜": date, "항목": item, "금액": amount})
    print(f"{item} 기록 완료!")

def show_total():
    total = sum(e["금액"] for e in expenses)
    print(f"현재까지 총 지출: {total}원")

# 실행 테스트
add_expense("2023-10-27", "점심밥", 9000)
add_expense("2023-10-27", "커피", 4500)
show_total()
```
